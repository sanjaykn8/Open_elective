```python
import numpy as np
import pandas as pd

from sklearn.datasets import load_iris
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import MinMaxScaler
from scipy.stats import zscore
```

```python
iris = load_iris(as_frame=True)
df = iris.frame
```

```python
df.iloc[5:10, 0] = np.nan

imputer = SimpleImputer(strategy='mean')
df[df.columns[:-1]] = imputer.fit_transform(df[df.columns[:-1]])
```

```python
minmax = MinMaxScaler()
df_minmax = df.copy()
df_minmax[df.columns[:-1]] = minmax.fit_transform(df[df.columns[:-1]])
```

```python
z = np.abs(zscore(df[df.columns[:-1]]))
df_no_outliers = df[(z < 3).all(axis=1)]
```

```python
df_encoded = pd.get_dummies(df, columns=['target'])
```
